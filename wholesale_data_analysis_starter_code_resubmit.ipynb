## Part 2: Transform the Data
Do we know that this client spent the more money than client 66037? If not, how would we find out? Transform the data using the steps below to prepare it for analysis.

```python
# Create a column that calculates the subtotal for each line using the unit_price and the qty
df['Subtotal'] = df['unit_price'] * df['qty']
df
```

```python
# Create a column for shipping price.
# Assume a shipping price of $7 per pound for orders over 50 pounds and $10 per pound for items 50 pounds or under.
def calculate_shipping_cost(order_df):
    total_weight = order_df['unit_weight'].sum()
    shipping_cost = 7 if total_weight > 50 else 10
    # Apply shipping cost per pound to total weight
    return shipping_cost * total_weight
# Group by order_id and apply the shipping cost calculation
shipping_costs = df.groupby('order_id').apply(calculate_shipping_cost).reset_index(name='shipping_price')
# Check if 'shipping_price' already exists in df
if 'shipping_price' in df.columns:
    # If it exists, drop the existing column to avoid duplicates
    df = df.drop(columns=['shipping_price'])
# Merge shipping costs back into the main DataFrame using a left merge to avoid duplicating rows
df = df.merge(shipping_costs, on='order_id', how='left') # use how='left' and on='order_id' for merging
df


```

```python
# Create a column for the total price using the subtotal and the shipping price along with a sales tax of 9.25%
df['total_price'] = (df['Subtotal'] + df['shipping_price']) * (1 + 0.0925)
df

```

```python
# Create a column for the cost of each line using unit cost, qty, and
# shipping price (assume the shipping cost is exactly what is charged to the client)
df['line_cost'] = df['unit_cost'] * df['qty'] + df['shipping_price']
df
```

```python
# Create a column for the profit of each line using line cost and line price
df['Profit'] = df['total_price'] - df['line_cost']
df


```

```python
# Format the data and rename the columns to names suitable for presentation.
column_renames = {
    'order_id': 'Order ID',
    'client_id': 'Client ID',
    'category': 'Category',
    'subcategory': 'Subcategory',
    'qty': 'Quantity',
    'unit_price': 'Unit Price',
    'unit_cost': 'Unit Cost',
    'unit_weight': 'Unit Weight',
    'Subtotal': 'Subtotal',
    'shipping_price': 'Shipping Price',
    'total_price': 'Total Price',
    'line_cost': 'Line Cost',
    'Profit': 'Profit'
    }



# Define the money columns.
money_columns = ['Subtotal', 'Shipping Price', 'Total Price', 'Line Cost', 'Profit']

# Define a function that converts a dollar amount to millions.
def currency_format_millions(amount):
  """
  Formats a dollar amount to millions of dollars.

  Args:
    amount: The dollar amount to format.

  Returns:
    The formatted amount in millions of dollars.
  """
  # Convert amount to numeric if it's a strin
  try:
    amount = float(amount)
  except ValueError:
    # Handle cases where conversion to float fails (e.g., non-numeric strings)
    return amount  # Or return a default value like 'N/A'
  return "${:,.2f}M".format(amount / 1000000)

# Rename the columns
df = df.rename(columns=column_renames)

# Rename the columns to reflect the change in the money format.
money_columns = ['Subtotal', 'Shipping Price', 'Total Price', 'Line Cost', 'Profit']

# Apply the currency_format_millions function to only the money columns.
formatted_values = df[money_columns].applymap(currency_format_millions)
display(formatted_values)



```
